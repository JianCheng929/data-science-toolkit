```python
###Part 1-Google Colab and Python
#1. Create a numpy array of size 2 x 2

#numpy is a python library that allows you to work with numbers and large datasets

import numpy as np
array_2x2 = np.array ([
    [1,2],
    [3,4]
])

print("This is my numpy 2x2 array")
print(array_2x2)
```

```python
#2. Create a random numpy array of size 3 x 4

#The first number is the row and second is the column
#So this is a grid of 3 rows and 4 columns
#This allows you to make an array of shape 3,4 and fill it with random numbers. Your numbers will be float
random3x4 = np.random.rand(3,4)

print("Randam 3x4 array")
print(random3x4)
```

```python
#3. Create a numpy array of ones and a numpy array of zeroes
ones_array = np.ones((3,3))
zeros_array = np.zeros((3,3))

print("Array of Ones: ")
print(ones_array)

print("Array of zeros:")
print(zeros_array)
```

```python
#4. Create a random numpy array of size 6x3 and transpose it

random6x3 = np.random.rand(6,3)
print("Original 6x3 Array:")
print(random6x3)

# Transpose the array (swap rows and columns)
transposed_array = random6x3.transpose()

print("\nTransposed 3x6 Array:")
print(transposed_array)

# Check the dimensions of both arrays
print("\nOriginal shape:", random6x3.shape)
print("Transposed shape:", transposed_array.shape)
```

```python
#5. Create two random numpy array of size n x p and p x m and then compute the dot product
# Set the dimensions
n = 2
p = 3
m = 4

# Generate two random arrays
array_A = np.random.rand(n, p)
array_B = np.random.rand(p, m)

print("Array A:")
print(array_A)

print("\nArray B:")
print(array_B)

# Compute the dot product
dot_product = np.dot(array_A, array_B)

print("\nDot Product:")
print(dot_product)

# Check the shapes
print("\nShape of Matrix A:", array_A.shape)
print("Shape of Matrix B:", array_B.shape)
print("Shape of Result:", dot_product.shape)
```
